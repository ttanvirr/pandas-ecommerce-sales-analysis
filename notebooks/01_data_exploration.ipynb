```python
# Challenge 1

from pathlib import Path;
import pandas as pd;

project_root = Path.cwd().parent
# print(project_root)
data_dir = project_root / "data" / "raw"

required_files = [
    "olist_orders_dataset.csv",
    "olist_order_items_dataset.csv", 
    "olist_customers_dataset.csv",
]

missing_files = [filename for filename in required_files if not (data_dir / filename).exists()]

if missing_files:
    raise FileNotFoundError(
        f"Missing files in {data_dir}: {missing_files}. "
        "Download and extract the Olist dataset into data/raw/."
    )

orders = pd.read_csv(data_dir / "olist_orders_dataset.csv")
orders_items = pd.read_csv(data_dir / "olist_order_items_dataset.csv")
customers = pd.read_csv(data_dir / "olist_customers_dataset.csv")

print("The three starting datasets were loaded successfully.")
```

```python
# Challenge 2

starting_datasets = {
    "orders": orders,
    "orders_items": orders_items,
    "customers": customers,
}

for dataset_name, dataframe in starting_datasets.items():
    print(f"\n--- {dataset_name} ---")
    print("Shape: ", dataframe.shape)
    print("Columns: ", list(dataframe.columns))
    dataframe.info()
    display(dataframe.head())
print("\n--- Numeric summary: order_items ---")
display(orders_items.describe())
```

```python
# Challenge 3

for dataset_name, dataframe in starting_datasets.items():
    print(f"\nMissing values in {dataset_name}:")
    display(dataframe.isna().sum().sort_values(ascending=False))

orders_duplicate_mask = orders.duplicated(subset=["order_id"], keep=False)
items_duplicate_mask = orders_items.duplicated(subset=["order_id", "order_item_id"], keep=False)
customers_duplicate_mask = customers.duplicated(subset=["customer_id"], keep=False)

print("Duplicate order id rows", int(orders_duplicate_mask.sum()))
print("Duplicate order id + order item id rows", int(items_duplicate_mask.sum()))
print("Duplicate customer id rows", int(customers_duplicate_mask.sum()))

if orders_duplicate_mask.any():
    display(orders.loc[orders_duplicate_mask].head())
if items_duplicate_mask.any():
    display(orders_items.loc[items_duplicate_mask].head())
if customers_duplicate_mask.any():
    display(customers.loc[customers_duplicate_mask].head())
```

```python
# Challenge 4

orders_clean = orders.copy()
order_items_clean = orders_items.copy()

order_date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
]

for column in order_date_columns:
    orders_clean[column] = pd.to_datetime(
        orders_clean[column],
        errors="coerce"
    )

order_items_clean["shipping_limit_date"] = pd.to_datetime(
    order_items_clean["shipping_limit_date"],
    errors="coerce"
)

print(orders_clean[order_date_columns].dtypes)
print("\nMissing converted order dates:")
display(orders_clean[order_date_columns].isna().sum())

print(order_items_clean["shipping_limit_date"].dtypes)
print("\nMissing shipping limit dates:")
display(order_items_clean["shipping_limit_date"].isna().sum())
```

```python
# Challenge 5

items_clean = order_items_clean.copy()

numeric_columns = ["price", "freight_value"]

for column in numeric_columns:
    items_clean[column] = pd.to_numeric(
        items_clean[column], errors="coerce"
    )

print("Missing or invalid price: ", items_clean["price"].isna().sum())
print("Missing or invalid freight value: ", items_clean["freight_value"].isna().sum())

items_clean["item_total_brl"] = (
    items_clean["price"] + items_clean["freight_value"]
)

suspicious_mask = (
    items_clean["price"].isna()
    | items_clean["freight_value"].isna()
    | items_clean["price"].lt(0)
    | items_clean["freight_value"].lt(0)
)

print("Suspicious item rows: ", int(suspicious_mask.sum()))
display(items_clean.loc[suspicious_mask].head())
display(items_clean[["price", "freight_value", "item_total_brl"]].describe())
```

```python
# Challenge 6: Summarise order statuses and order-item values

## Count orders by `order_status` and sort the result from largest to smallest.
order_status_counts = orders_clean["order_status"].value_counts().sort_values(ascending=False)
display(order_status_counts.reset_index())

## Group item lines by `order_id`
order_item_summary = (
    items_clean
    .groupby("order_id")
    .agg(
        product_sales_brl=("price", "sum"),
        freight_brl=("freight_value", "sum"),
        item_line_count=("order_item_id", "count")
    )
    .reset_index()
    .sort_values(by="product_sales_brl", ascending=False)
)


orders_with_item_summary = orders_clean.merge(
    order_item_summary,
    on="order_id",
    how="left",
    validate="one_to_one"
).sort_values(by="product_sales_brl", ascending=False)

print("Orders before merge: ", len(orders_clean))
print("Rows after merge: ", len(orders_with_item_summary))
display(order_item_summary.head())
display(orders_with_item_summary.head())

## For each order, calculate the sum of price, the sum of freight_value, and the number of item lines.
```

```python
# Challenge 7: Analyse monthly completed-order sales

## Select only orders whose order_status is delivered
delivered_order_dates = orders_clean.loc[
    orders_clean["order_status"] == "delivered",
    ["order_id", "order_purchase_timestamp"],
].copy()
# display(delivered_order_dates.head())

## Merge those rows with `items_clean` on `order_id`.
completed_items = items_clean.merge(
    delivered_order_dates,
    on="order_id",
    how="inner",
    validate="many_to_one"
)
# display(completed_items.head())

## Create a `purchase_month` column from `order_purchase_timestamp` using monthly periods
completed_items["purchase_month"] = (
    completed_items["order_purchase_timestamp"]
    .dt.to_period("M")
)
# display(completed_items.head())

## Group by month and calculate product price total, freight total, combined item-plus-freight total, distinct order count, and item-line count
monthly_sales_summary = (
    completed_items
    .groupby("purchase_month")
    .agg(
        product_sales_brl=("price", "sum"),
        freight_brl=("freight_value", "sum"),
        item_plus_freight_brl=("item_total_brl", "sum"),
        distinct_order_count=("order_id", "nunique"),
        item_line_count=("order_item_id", "count")
    )
    .sort_index()
)

display(monthly_sales_summary)
print("Month with the highest product sales:",
monthly_sales_summary["product_sales_brl"].idxmax()
)
```

```python
# Challenge 8: Find the strongest product categories

## Load the products and category-translation CSV files
products = pd.read_csv(data_dir / "olist_products_dataset.csv")
category_translation = pd.read_csv(data_dir / "product_category_name_translation.csv")

## Merge `items_clean` with products using `product_id`.
items_with_products = items_clean.merge(
    products,
    on="product_id",
    how="left",
    validate="many_to_one"
)

## Merge the result with `category_translation` using `product_category_name`.
items_with_categories = items_with_products.merge(
    category_translation,
    on="product_category_name",
    how="left",
    validate="many_to_one"
)

## Use the English category name when available. 
# If it is missing, keep the original Portuguese category name. 
# If both are missing, use `Uncategorized`.
items_with_categories["category_name"] = (
    items_with_categories["product_category_name_english"]
    .fillna(items_with_categories["product_category_name"])
    .fillna("Uncategorized")
)

category_summary = (
    items_with_categories
    .groupby("category_name")
    .agg(
        product_sales_brl=("price", "sum"),
        item_line_count=("order_item_id", "count"),
        distinct_product=("product_id", "nunique")
    )
    .sort_values(by="product_sales_brl", ascending=False)
)

display(category_summary.head(10))
```

```python
# Challenge 9: Compare orders and customers by state

delivered_orders = orders_clean.loc[
    orders_clean["order_status"] == "delivered",
].copy()

delivered_orders_with_customers = delivered_orders.merge(
    customers[
        [
            "customer_id",
            "customer_unique_id",
            "customer_state",
        ]
    ],
    on="customer_id",
    how="left",
    validate="many_to_one"
)

state_summary = (
    delivered_orders_with_customers
    .groupby("customer_state")
    .agg(
        distinct_order_count=("order_id", "nunique"),
        distinct_customer_count=("customer_unique_id", "nunique")
    )
    .sort_values(by="distinct_order_count", ascending=False)
)
display(state_summary)
```
